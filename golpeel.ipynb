```python
# Troca o "\n" que ficou digitado errado (como texto) por um espaço em branco
df['texto'] = df['texto'].str.replace(r'\\n', ' ', regex=True)
```

```python
# Troca espaços duplicados/triplicados por um espaço só, e remove espaços do início/fim da frase
df['texto'] = df['texto'].str.replace(r'\s+', ' ', regex=True).str.strip()
```
