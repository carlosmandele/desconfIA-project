```python
# Biblioteca que permite encontrar padrões dentro de textos
import re

# Função de Tokenização
def tokenizar(texto):
    # findall() vai procurar todas as ocorrências que correspondem ao padrão definido pelo Regex
    return re.findall(r'\b\w+\b', texto.lower())

# Aplica a função tokenizar em cada valor da coluna texto
df['tokens'] = df['texto'].apply(tokenizar)

df[['texto', 'tokens']].head()
```

```python
import torch
from transformers import AutoTokenizer, AutoModel
```

```python
# Nome do modelo de IA pré-treinado que vamos utilizar.
# Esse modelo é o DistilBERT, preparado para trabalhar com vários idiomas.
modelo_nome = "distilbert-base-multilingual-cased"

# Carrega o tokenizer correspondente ao modelo escolhido.
tokenizer = AutoTokenizer.from_pretrained(modelo_nome)

# Carrega o modelo DistilBERT pré-treinado.
modelo = AutoModel.from_pretrained(modelo_nome)
```

```python
# Pega o primeiro texto da coluna 'texto' do DataFrame.
texto = df['texto'].iloc[0]

# Passa o texto pelo tokenizer do DistilBERT.
entrada = tokenizer(
    texto,
    return_tensors="pt",
    truncation=True,
    padding=True
)

entrada
```

```python
# Executa o modelo sem calcular gradientes.
# Como neste momento estamos apenas usando o modelo pré-treinado
# para obter uma representação do texto, não precisamos treinar o modelo.
with torch.no_grad():

    # Envia a entrada gerada pelo tokenizer para o DistilBERT.
    saida = modelo(**entrada)

# Pega a representação do primeiro token de cada mensagem.
embedding = saida.last_hidden_state[:, 0, :]

# Mostra o formato (dimensões) do embedding gerado.
print(embedding.shape)
```

```python
import numpy as np

# Cria uma função para gerar embeddings a partir de vários textos.
# batch_size=16 significa que vamos processar 16 mensagens por vez.
def gerar_embeddings(textos, batch_size=16):
    embeddings = []

    for i in range(0, len(textos), batch_size):

        # Seleciona um grupo de até 16 textos para ser processado.
        batch = textos[i:i + batch_size]

        # Usa o tokenizer do DistilBERT para transformar
        # as mensagens do grupo em entradas que o modelo entende.
        entradas = tokenizer(
            batch,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=128
        )

        # Executa o modelo sem calcular gradientes,
        # pois estamos apenas gerando embeddings e não treinando o modelo.
        with torch.no_grad():
            saida = modelo(**entradas)

        # Pega a representação do primeiro token de cada mensagem
        embedding = saida.last_hidden_state[:, 0, :]

        embeddings.append(embedding.cpu().numpy())

    return np.vstack(embeddings)
```

```python
# Pega todos os textos da coluna 'texto' do DataFrame
# e transforma esses textos em uma lista.
# Depois, essa lista é enviada para a função gerar_embeddings().
X_embeddings = gerar_embeddings(df['texto'].tolist())

# Mostra o formato da matriz de embeddings.
print("Formato dos embeddings:", X_embeddings.shape)
```
